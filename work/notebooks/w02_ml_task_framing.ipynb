## 5. Why ML beats a fixed rule here

```python
Fixed rules like "posts with more hashtags get boosted" or "new posts always rank high" are brittle. 
They break when trends change or users game the system. 
ML learns patterns from text, author history, time of day, and engagement signals. 
It adapts automatically to new slang, memes, and creator behavior. Rules can't scale, ML can.
```
