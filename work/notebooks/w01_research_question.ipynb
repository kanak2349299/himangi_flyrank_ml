## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
df_filtered=df[df["impressions_90d"]>0]
corr = df_filtered["search_volume"].corr(df_filtered["impressions_90d"])
print(f"Correlation between search_volume and impressions_90d: {corr:.3f}")
if corr>0.3:
  print("Observed :Moderate positive correlation.Higher search volume relates to higher impressions.")
elif corr>0:
  print("Observed:Weak positive correlation.Search volume has slight relation with impressions.")
else:
  print("Observed:No correlation.Search volume has no relation with impressions.")

'''observation:
After filtering the pages with "impressions_90d">0,correlation between search_volume and impressions_90d has decreased to 0.001.'''
```
